**AUTOGRAD:**

module to calculate the gradient of any function automatically
(IMPORTANT IN BACK PROPOGATION)

Example 1:
y=x^2

---

Example 2:
y=x**2;
z=sin(y)


```python
import torch
import numpy as np
```

```python
x=torch.tensor(3.0,requires_grad=True)
```

Requires_grad:

tell torch that we will be needing this tensors depending on it their gradient gradient and keep the track of all the function on it and torch builds it tree/computation graph


```python
y=x**2
```

```python
x
```

```python
y
```

```python
z=torch.sin(y)
```

```python
z
```

calculating derivative of y w.r.t x

```python
if x.grad is not None:
    x.grad.zero_()
y.backward(retain_graph=True)
print(x.grad)
```

IMPORTANT

---
1. after the backward function is applied on a tensor in the graph pytorch traverses from the intiall node(X) to the leaf node and find the derivative and the memory for the graph by default is freed for optimization

2.by using retain_graph=true :the memory/graph is retained but the backward can't be applied on same tensor

ABOUT COMPUTATIONAL GRAPHS:

1.  LEAF NODE: INPUT NODE(X)
2.  ROOT NODE:OUTPUT NODE(Y)

THE .grad is only calculated for the leaf node





```python
if x.grad is not None:
    x.grad.zero_() # Clear previous gradients
z.backward(retain_graph=True)
print(x.grad)
```
