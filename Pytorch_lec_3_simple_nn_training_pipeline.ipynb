```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

Processing data

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python
df=df.drop(columns=["Unnamed: 32","id"])
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

```python
X_train
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
X_train
```

```python
y_train
```

```python
X_train_tensor=torch.from_numpy(X_train)
X_test_tensor=torch.from_numpy(X_test)
y_train_tensor=torch.from_numpy(y_train)
y_test_tensor=torch.from_numpy(y_test)
```

```python
X_train_tensor
```

Building Model

```python
class Nueralnetwork():

  def __init__(self,X):

    self.weights=torch.randn(X.shape[1],1,dtype=torch.float64,requires_grad=True)
    self.bias=torch.zeros(1,dtype=torch.float64,requires_grad=True)

  def forward(self,X):
    z=torch.sigmoid(torch.matmul(X,self.weights)+self.bias)
    y_pred=torch.sigmoid(z)
    return y_pred

  def loss(self,y_pred,y):
    # Clamp predictions to avoid log(0)
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)

    # Calculate loss
    loss = -(y * torch.log(y_pred) + (1 - y) * torch.log(1 - y_pred)).mean()
    return loss
```

```python
learning_rate=0.1
epochs=25
```

```python
model=Nueralnetwork(X_train_tensor)
```

```python
model.weights
```

training pipeline

```python
for epoch in range(epochs):
  #forward pass
  y_pred=model.forward(X_train_tensor)
  #loss calculation
  loss=model.loss(y_pred,y_train_tensor)
  print(epoch,"loss:",loss)
  #backward pass
  loss.backward()
  #updation
   #off.no grad during upadation
  with torch.no_grad():
    model.weights-=learning_rate*model.weights.grad
    model.bias-=learning_rate*model.bias.grad
    #on grad.zero to stop accumlation
    model.weights.grad.zero_()
    model.bias.grad.zero_()

    print(epoch," weights:", model.weights)
    print(epoch,"bais:",model.bias)
```

test

```python
with torch.no_grad():
  y_pred=model.forward(X_test_tensor)
  loss=model.loss(y_pred,y_test_tensor)
  print("loss:",loss)
```
